```python
# Merge orders with order items
sales_data = orders.merge(
    order_items,
    on="order_id",
    how="inner"
)

# Keep only delivered orders
sales_data = sales_data[
    sales_data["order_status"] == "Delivered"
].copy()

# Calculate net revenue
sales_data["revenue"] = (
    sales_data["quantity"]
    * sales_data["unit_price"]
    * (1 - sales_data["discount_percent"] / 100)
)

# Create month column
sales_data["month"] = sales_data["order_date"].dt.to_period("M")

# Monthly revenue
monthly_revenue = (
    sales_data.groupby("month")["revenue"]
    .sum()
    .round(2)
    .reset_index()
)

monthly_revenue
```

```python
# RFM Analysis

rfm = sales_data.groupby("customer_id").agg(
    recency=("order_date", lambda x: (sales_data["order_date"].max() - x.max()).days),
    frequency=("order_id", "nunique"),
    monetary=("revenue", "sum")
).reset_index()

rfm["monetary"] = rfm["monetary"].round(2)

rfm.head(10)
```

```python
# RFM Scoring

rfm["R_score"] = pd.qcut(
    rfm["recency"],
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_score"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_score"] = pd.qcut(
    rfm["monetary"],
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_score"] = (
    rfm["R_score"].astype(str)
    + rfm["F_score"].astype(str)
    + rfm["M_score"].astype(str)
)

rfm.head(10)
```

```python
def assign_segment(row):
    if row["R_score"] >= 4 and row["F_score"] >= 4 and row["M_score"] >= 4:
        return "Champions"
    elif row["R_score"] >= 3 and row["F_score"] >= 3:
        return "Loyal Customers"
    elif row["R_score"] >= 4 and row["F_score"] <= 3:
        return "Potential Loyalists"
    elif row["R_score"] <= 2 and row["F_score"] >= 3:
        return "At Risk"
    else:
        return "Lost Customers"


rfm["customer_segment"] = rfm.apply(assign_segment, axis=1)

rfm[[
    "customer_id",
    "RFM_score",
    "customer_segment"
]].head(10)
```

```python
rfm_summary = (
    rfm.groupby("customer_segment")
    .agg(
        customer_count=("customer_id", "count"),
        total_revenue=("monetary", "sum")
    )
    .sort_values("total_revenue", ascending=False)
    .round(2)
    .reset_index()
)

rfm_summary
```

```python
discount_analysis = sales_data.copy()

discount_analysis["discount_range"] = pd.cut(
    discount_analysis["discount_percent"],
    bins=[-1, 0, 10, 20, 30, 40, 50],
    labels=[
        "No Discount",
        "1-10%",
        "11-20%",
        "21-30%",
        "31-40%",
        "41-50%"
    ]
)

discount_summary = (
    discount_analysis
    .groupby("discount_range", observed=False)
    .agg(
        total_orders=("order_id", "nunique"),
        total_revenue=("revenue", "sum")
    )
    .round(2)
    .reset_index()
)

discount_summary
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    discount_summary["discount_range"].astype(str),
    discount_summary["total_revenue"]
)

plt.title("Revenue by Discount Range")
plt.xlabel("Discount Range")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
profit_data = sales_data.merge(
    products[["product_id", "category", "cost_price", "selling_price"]],
    on="product_id",
    how="left"
)

profit_data["profit"] = (
    profit_data["quantity"]
    * (
        profit_data["selling_price"]
        - profit_data["cost_price"]
    )
    * (1 - profit_data["discount_percent"] / 100)
)

category_profit = (
    profit_data
    .groupby("category")
    .agg(
        total_revenue=("revenue", "sum"),
        total_profit=("profit", "sum")
    )
    .round(2)
    .sort_values("total_profit", ascending=False)
    .reset_index()
)

category_profit
```

```python
plt.figure(figsize=(11, 6))

x = range(len(category_profit))

plt.bar(
    [i - 0.2 for i in x],
    category_profit["total_revenue"],
    width=0.4,
    label="Revenue"
)

plt.bar(
    [i + 0.2 for i in x],
    category_profit["total_profit"],
    width=0.4,
    label="Profit"
)

plt.title("Revenue vs Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(x, category_profit["category"], rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
